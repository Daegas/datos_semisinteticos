```python
# el archivo está mal códigificado, por lo que se debe reparar el mojibake
def reparar_mojibake(texto):
    if isinstance(texto, str):
        try:
            return texto.encode("latin1").decode("utf-8")
        except (UnicodeEncodeError, UnicodeDecodeError):
            return texto
    return texto
```

## LOCALIDAD

```python
def agregar_localidad(df, mapeo_lugar, tipo):
    """
    Agrega el nombre de la localidad usando las claves
    entidad + municipio + localidad del catálogo INEGI.
    """

    ent = f"ent_{tipo}"
    mun = f"mun_{tipo}"
    loc = f"loc_{tipo}"
    loc_nom = f"{loc}_nom"

    mapeo_loc = mapeo_lugar[
        ["cve_ent", "cve_mun", "cve_loc", "nom_loc"]
    ].copy()

    # Asegurar el mismo tipo de dato en las claves
    df[ent] = df[ent].astype(str)
    df[mun] = df[mun].astype(str)
    df[loc] = df[loc].astype(str)

    mapeo_loc["cve_ent"] = mapeo_loc["cve_ent"].astype(str)
    mapeo_loc["cve_mun"] = mapeo_loc["cve_mun"].astype(str)
    mapeo_loc["cve_loc"] = mapeo_loc["cve_loc"].astype(str)

    mapeo_loc.columns = [ent, mun, loc, loc_nom]

    df = pd.merge(
        df,
        mapeo_loc,
        on=[ent, mun, loc],
        how="left"
    )

    return df
```

```python
year_data_fixed = agregar_localidad(
    year_data_fixed, df_catalogo, "regis"
)

year_data_fixed = agregar_localidad(
    year_data_fixed, df_catalogo, "resid"
)

year_data_fixed = agregar_localidad(
    year_data_fixed, df_catalogo, "ocurr"
)

year_data_fixed = agregar_localidad(
    year_data_fixed, df_catalogo, "ocules"
)
```

```python
import unicodedata

def _normalizar_nombre_col(col):
    # quitar acentos y pasar a minúsculas
    col = str(col).strip().lower()
    col = unicodedata.normalize("NFKD", col).encode("ascii", "ignore").decode("ascii")
    return col


def cargar_catalogo(nombre_catalogo, year_datapath=YEAR_DATAPATH):
    """
    Carga un catálogo del INEGI por nombre (sin .csv).
    - Repara mojibake en columnas string.
    - Normaliza nombres de columnas a lowercase (sin acentos/espacios).
    """
    catalogo_path = year_datapath / "catalogos" / f"{nombre_catalogo}.csv"
    df_cat = pd.read_csv(catalogo_path, low_memory=False)


    # reparar mojibake en columnas string
    for col in df_cat.select_dtypes(include=["object", "str"]).columns:
        df_cat[col] = df_cat[col].map(reparar_mojibake)
        
    # renombrar columnas a lowercase normalizado
    df_cat.columns = [_normalizar_nombre_col(c) for c in df_cat.columns]

    return df_cat
```

```python
def manual_fixes(df, col, cat):
    """
    Aplica correcciones manuales a los DataFrames según la columna a decodificar.
    """
    if col == "lista1":
        # Convertir a Int64 para manejar NaN
        df[col] = df[col].astype("Int64")
        cat.iloc[:, 0] = cat.iloc[:, 0].astype("Int64")
    elif col == "gr_lismex":
        # Asegurar que los códigos tengan dos dígitos con ceros a la izquierda
        df[col] = df[col].astype(str).str.zfill(2)
        cat.iloc[:, 0] = cat.iloc[:, 0].astype(str).str.zfill(2)
    return cat, df

def decodificar_normal(df, col, nombre_catalogo, col_clave=None, col_valor=None):
    """
    Decodifica una columna usando un catálogo.
    - df: DataFrame
    - col: nombre de la columna a decodificar (ej. 'sexo')
    - nombre_catalogo: nombre del catálogo sin .csv
    - col_clave: columna del catálogo con la clave (autodetecta si None)
    - col_valor: columna del catálogo con el valor legible (autodetecta si None)
    
    Retorna el df con una nueva columna '<col>_nom'.
    """
    cat = cargar_catalogo(nombre_catalogo)

    cat, df = manual_fixes(df, col, cat)
    
    # Autodetectar columnas: la clave suele ser la primera, el valor la última
    if col_clave is None:
        col_clave = cat.columns[0]
    if col_valor is None:
        try:
            col_valor = next(c for c in cat.columns if c.lower() == "descrip")
        except KeyError:
            col_valor = cat.columns[-1]
    print(f"    col_clave: {col_clave}, col_valor: {col_valor}")
    
    # Asegurar tipos compatibles
    cat[col_clave] = cat[col_clave].astype(str)
    df[col] = df[col].astype(str)
    df[col] = df[col].str.strip()
    
    mapeo = dict(zip(cat[col_clave], cat[col_valor]))
    df[f"{col}_nom"] = df[col].map(mapeo)
    return df
```
