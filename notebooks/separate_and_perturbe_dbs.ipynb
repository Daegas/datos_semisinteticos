## Nombres cortos o sin vocales

```python
# eliminar registros AEIOU
df_len = len(df)
df = df[df["apellido_paterno"].str.contains(r"[AEIOUY]", na=False)]
print(f"Filas eliminadas por apellido_paterno sin vocales: {df_len - len(df)}")

df_len = len(df)
df = df[df["apellido_materno"].str.contains(r"[AEIOUY]", na=False)]
print(f"Filas eliminadas por apellido_materno sin vocales: {df_len - len(df)}")


df_len = len(df)
df = df[df["nombre"].str.contains(r"[AEIOUY]", na=False)]
print(f"Filas eliminadas por nombre sin vocales: {df_len - len(df)}")  
```

```python
# Eliminar registros con un solo caracter
df_len = len(df)
df = df[df["apellido_paterno"].str.len() != 1]
print(f"Filas eliminadas por apellido_paterno de una letra: {df_len - len(df)}")

df_len = len(df)
df = df[df["apellido_materno"].str.len() != 1]
print(f"Filas eliminadas por apellido_materno de una letra: {df_len - len(df)}")

df_len = len(df)
df = df[df["nombre"].str.len() != 1]
print(f"Filas eliminadas por nombre de una letra: {df_len - len(df)}")
```

```python
# Eliminar registros que terminan en punto
df_len = len(df)
df = df[~df["apellido_paterno"].str.endswith(".", na=False)]
print(f"Filas eliminadas por apellido_paterno terminando en punto: {df_len - len(df)}")
df_len = len(df)
df = df[~df["apellido_materno"].str.endswith(".", na=False)]
print(f"Filas eliminadas por apellido_materno terminando en punto: {df_len - len(df)}")
df_len = len(df)
df = df[~df["nombre"].str.endswith(".", na=False)]
print(f"Filas eliminadas por nombre terminando en punto: {df_len - len(df)}")
```

# SEPARAR 

```python
set_a_columns = ["sexo", "cve_lengua_nom", "nacesp_cve_nom", "fecha_nacim", "nombre", "apellido_paterno", "apellido_materno", "ent_resid_nom", "tloc_resid_nom", "tipo_defun_nom", "par_agre_nom", "vio_fami_nom", "cirugia_nom", "natviole_nom"]
set_b_columns = ["sexo", "ent_nac_nom", "afromex_nom", "conindig_nom", "nacionalid_nom", "edad", "ocupacion_nom", "edad_agru", "nombre", "apellido_paterno", "apellido_materno", "ent_resid_nom", "mun_resid_nom", "loc_resid_nom", "fecha_regis", "causa_def_nom", "encefalica_nom", "rel_emba_nom", "maternas_nom", "razon_m_nom"]
set_c_columns = ["sexo", "lengua_nom", "edad", "cond_act_nom", "edad_agru", "nombre", "apellido_paterno", "apellido_materno", "mun_resid_nom", "area_ur_nom", "lista_mex_nom", "ent_ocurr_nom", "mun_ocurr_nom", "loc_ocurr_nom", "fecha_ocurr"]
```

```python
import numpy as np



from perturbation import (
    add_person_id,
    add_duplicates,
    drop_individuals,
    perturb_dataframe,
    SPEC_A,
    SPEC_B,
    SPEC_C
)

SEED = 2027
P_RECORD = 0.6   # prob. de que un registro sea perturbado
P_COLUMN = 0.2   # prob. de perturbar cada columna dentro de un registro elegido
MAPPING_DIR = DATAPATH / "diccionarios" / "sinonimos"

DATASETS = {
    "A": dict(cols=set_a_columns, spec=SPEC_A,
              p_drop=0.05, p_dup=0.05, max_copies=1),

    "B": dict(cols=set_b_columns, spec=SPEC_B,
              p_drop=0.15, p_dup=0.10, max_copies=2),

    "C": dict(cols=set_c_columns, spec=SPEC_C,
              p_drop=0.10, p_dup=0.02, max_copies=1),
}

df = add_person_id(df)

results, reports = {}, {}

for i, (name, cfg) in enumerate(DATASETS.items()):

    rng = np.random.default_rng([SEED, i])

    base = df[["id_persona"] + cfg["cols"]].copy()

    # 1. Eliminar individuos
    n_before = base["id_persona"].nunique()

    base = drop_individuals(
        base,
        cfg["p_drop"],
        rng=rng
    )

    n_after = base["id_persona"].nunique()

    print(
        f"\n=== Dataset {name} ==="
        f"\nIndividuos eliminados: {n_before - n_after:,}"
        f"\nIndividuos restantes: {n_after:,}"
    )

    # 2. Generar duplicados
    base = add_duplicates(
        base,
        cfg["p_dup"],
        rng=rng,
        max_copies=cfg["max_copies"]
    )

    print(f"Filas después de duplicar: {len(base):,}")

    # 3. Perturbar registros
    results[name], reports[name] = perturb_dataframe(
        base,
        cfg["spec"],
        P_RECORD,
        P_COLUMN,
        mapping_dir=MAPPING_DIR,
        seed=[SEED, i, 1],
    )

for name, d in results.items():
    path = DATAPATH / "cleaned" / "datasets"
    path.mkdir(parents=True, exist_ok=True)
    d.to_parquet(path / f"dataset_{name}.parquet", index=False)
    d.sample(n=1000, random_state=SEED).to_csv(path / f"dataset_{name}_sample.csv", index=False)
```
