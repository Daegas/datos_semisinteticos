## Funciones

### Perturbación de columnas categóricas

La función `perturb_column_dic` permite introducir variaciones controladas en los valores de una columna de un DataFrame utilizando un diccionario de mapeo (`mapping`). Su objetivo es generar diferentes representaciones de una misma categoría, simulando variaciones que pueden aparecer en distintas fuentes de datos.

Por ejemplo, una misma entidad geográfica puede aparecer registrada con diferentes nombres, abreviaturas o variaciones ortográficas:

```python
mapping = {
    "Nuevo León": [
        "Nuevo León",
        "Nuevo Leon",
        "NL",
        "Nuevo León, México"
    ]
}
```

La función selecciona aleatoriamente una de las representaciones disponibles para cada registro, controlando la probabilidad de conservar el valor original mediante el parámetro `temperature`.

#### Parámetros

| Parámetro | Descripción | Valor por defecto |
|---|---|---|
| `df` | DataFrame que contiene los registros. | Requerido |
| `column` | Nombre de la columna que se desea perturbar. | Requerido |
| `mapping` | Diccionario que relaciona cada categoría original con sus representaciones alternativas. | Requerido |
| `temperature` | Controla la probabilidad relativa de seleccionar una alternativa frente a conservar el valor original. | `1.0` |
| `seed` | Semilla para reproducibilidad del muestreo aleatorio. | `None` |
| `inplace` | Indica si la modificación se realiza sobre el DataFrame original. | `False` |

#### Temperatura y distribución de probabilidad

El parámetro `temperature` controla la intensidad de las perturbaciones. La categoría original recibe un peso de 1, mientras que cada alternativa recibe un peso igual a `temperature`. Posteriormente, los pesos se normalizan para obtener una distribución de probabilidad.

Sea $k$ el número total de representaciones disponibles para una categoría, incluyendo la original, y sea $T$ el valor de `temperature`. La probabilidad de seleccionar cada representación está dada por:

$$
P(x_i) =
\begin{cases}
\frac{1}{1+(k-1)T}, & x_i = x_{\text{original}} \\[6pt]
\frac{T}{1+(k-1)T}, & x_i \neq x_{\text{original}}
\end{cases}
$$

La interpretación del parámetro es la siguiente:

| Temperature | Comportamiento |
|---|---|
| `0.0` | Conserva siempre la categoría original. |
| `0.25` | Favorece considerablemente la categoría original. |
| `0.5` | Favorece moderadamente la categoría original. |
| `1.0` | Todas las representaciones tienen la misma probabilidad. |
| `2.0` | Favorece las representaciones alternativas. |

Por ejemplo, considerando una categoría con cuatro representaciones y `temperature=0.5`, se obtiene:

| Representación | Probabilidad |
|---|---:|
| Nuevo León (original) | 40% |
| Nuevo Leon | 20% |
| NL | 20% |
| Nuevo León, México | 20% |

#### Consideraciones

- Las alternativas duplicadas se eliminan antes del muestreo, conservando el orden original.
- Si una categoría no está presente en el diccionario de mapeo, su valor permanece sin cambios.
- Si una categoría solamente tiene una representación disponible, se conserva su valor.
- La selección se realiza mediante muestreo aleatorio independiente para cada registro.


````python
import json
import numpy as np


def load_mapping(json_path: str | Path) -> dict[str, list[str]]:
    with open(json_path, "r", encoding="utf-8") as f:
        return json.load(f)


def perturb_column_dic(
    df: pd.DataFrame,
    column: str,
    mapping: dict[str, list[str]],
    *,
    temperature: float = 1.0,
    seed: int | None = None,
    inplace: bool = False,
) -> pd.DataFrame:

    if temperature < 0:
        raise ValueError("temperature debe ser >= 0")

    if not inplace:
        df = df.copy()

    rng = np.random.default_rng(seed)

    values = df[column].to_numpy()

    cats = list(mapping)
    cat_to_idx = {c: i for i, c in enumerate(cats)}

    alternatives = [
        list(dict.fromkeys([c] + mapping[c]))
        for c in cats
    ]

    row_idx = (
        df[column]
        .map(cat_to_idx)
        .fillna(-1)
        .to_numpy(dtype=np.int64)
    )

    mask = row_idx >= 0
    out = values.copy()

    for i, alts in enumerate(alternatives):

        rows = np.flatnonzero(row_idx == i)

        if len(rows) == 0:
            continue

        n = len(alts)

        if n == 1 or temperature == 0:
            continue

        weights = np.full(n, temperature, dtype=float)
        weights[0] = 1.0

        probabilities = weights / weights.sum()

        selected = rng.choice(
            n,
            size=len(rows),
            p=probabilities
        )

        out[rows] = np.asarray(alts, dtype=object)[selected]

    df[column] = out

    return df
````

### Perturbación de columnas numéricas


La perturbación de variables numéricas permite introducir pequeñas variaciones aleatorias en los valores de una columna, simulando errores de captura o modificaciones en los datos originales.


#### Parámetros

| Parámetro | Tipo | Default | Descripción |
|---|---|---|---|
| `df` | `pd.DataFrame` | Obligatorio | DataFrame que contiene los registros. |
| `column` | `str` | Obligatorio | Nombre de la columna que se desea perturbar. |
| `probability` | `float` | `1.0` | Probabilidad de aplicar ruido a cada registro. Debe estar entre 0 y 1. |
| `noise_range` | `tuple[int, int] \| None` | `(0, 5)` | Rango de magnitud del ruido que se agregará. Si es `None`, no se aplica ruido numérico. |
| `seed` | `int \| None` | `None` | Semilla para reproducibilidad de los resultados. |
| `inplace` | `bool` | `False` | Si es `True`, modifica el DataFrame original. |

#### Funcionamiento

Para cada registro se determina aleatoriamente si recibirá una perturbación, de acuerdo con `probability`.

Cuando se selecciona un registro, se genera una magnitud aleatoria dentro de `noise_range` y se determina aleatoriamente si se suma o se resta al valor original.

El ruido se expresa como:

$$
\epsilon_i = s_i \cdot u_i
$$

Donde:

- $s_i \in \{-1, 1\}$ determina el signo del ruido.
- $u_i \sim U(a,b)$ determina su magnitud.
- $(a,b)$ corresponde al rango definido en `noise_range`.

El valor perturbado se obtiene mediante:

$$
x_i' = x_i + \epsilon_i
$$

Los valores faltantes no se modifican.

### Consideraciones

- `probability = 0` no modifica ningún registro.
- `probability = 1` considera todos los registros para aplicar ruido.
- `noise_range = (0, 5)` permite variaciones de hasta cinco unidades en ambas direcciones.
- Si se utiliza un rango que incluye cero, algunos registros seleccionados podrían conservar su valor original.
- Para variables como edad, es recomendable establecer un límite inferior de cero para evitar valores negativos.

````python
def perturb_numeric_column(
    df: pd.DataFrame,
    column: str,
    *,
    probability: float = 1.0,
    noise_range: tuple[int, int] = (0, 5),
    seed: int | None = None,
    inplace: bool = False,
) -> pd.DataFrame:
    print("Probability:", probability)
    if not 0 <= probability <= 1:
        raise ValueError("probability debe estar entre 0 y 1")

    min_noise, max_noise = noise_range

    if min_noise < 0 or max_noise < min_noise:
        raise ValueError("noise_range debe ser un rango válido y no negativo")

    if not inplace:
        df = df.copy()

    rng = np.random.default_rng(seed)

    values = pd.to_numeric(df[column], errors="coerce").to_numpy(dtype=float)

    # Seleccionar registros que recibirán ruido
    mask = rng.random(len(values)) < probability

    # Generar magnitud y dirección del ruido
    magnitude = rng.integers(
        min_noise,
        max_noise + 1,
        size=len(values)
    )

    sign = rng.choice([-1, 1], size=len(values))

    noise = magnitude * sign

    # Evitar modificar valores faltantes
    mask &= ~np.isnan(values)

    out = values.copy()
    out[mask] += noise[mask]

    df[column] = out

    return df
````

````python
def perturb_and_report(
    df_: pd.DataFrame,
    perturbation_type: str,
    column: str,
    mapping_name: str | None = None,
    noise_range: tuple[int, int] | None = (0, 5),
    *,
    temperature: float = 1.0,
    seed: int | None = None,
    plot: bool = False,
    inplace: bool = False,
) -> pd.DataFrame:

    print(f"Perturbando columna: {column}")

    original = df_[column].copy()
    if plot:
        df_[column].value_counts().plot(kind="bar", color="red")



    if perturbation_type == "categorical":
        if mapping_name is None:
            raise ValueError("mapping_name debe ser proporcionado para perturbaciones categóricas")
        print(f"    Mapping: {mapping_name}")
        mapping_ = load_mapping(
            DATAPATH / "diccionarios" / "sinonimos" / f"{mapping_name}.json"
        )
        df_ = perturb_column_dic(
            df_,
            column=column,
            mapping=mapping_,
            temperature=temperature,
            seed=seed,
            inplace=inplace
        )
    elif perturbation_type == "numeric":
        df_ = perturb_numeric_column(
            df_,
            column=column,
            probability=temperature,
            noise_range=noise_range,
            seed=seed,
            inplace=inplace
        )

    changed = original.ne(df_[column])
    n_changed = changed.sum()
    pct_changed = changed.mean() * 100

    unique_before = original.nunique()
    unique_after = df_[column].nunique()


    print(f"    Registros modificados: {n_changed:,}")
    print(f"    Porcentaje modificado: {pct_changed:.2f}%")
    print(f"    Valores únicos: {unique_before} -> {unique_after}")

    if plot:
        plt.figure(figsize=(10, 5))
        df_[column].value_counts().plot(kind="bar", color="skyblue")

        plt.title(f"Distribución de frecuencias: {column}")
        plt.ylabel("Frecuencia")
        plt.xlabel("Categorías")
        plt.tight_layout()
        plt.show()

    return df_
````

### Perturbación de Fechas

````python
def perturb_datetime_column(
    df: pd.DataFrame,
    column: str,
    *,
    probability: float = 1.0,
    seed: int | None = None,
    inplace: bool = False,
) -> pd.DataFrame:

    if not 0 <= probability <= 1:
        raise ValueError("probability debe estar entre 0 y 1")

    if not inplace:
        df = df.copy()

    rng = np.random.default_rng(seed)

    # 1. Convertir a datetime
    dates = pd.to_datetime(df[column], errors="coerce")
    out = dates.copy()

    # 2. Seleccionar registros que recibirán ruido
    mask = rng.random(len(df)) < probability

    # No modificar fechas faltantes o inválidas
    mask &= dates.notna().to_numpy()

    # 3. Elegir componente y dirección del ruido
    components = rng.choice(
        ["days", "months", "years"],
        size=len(df)
    )
    signs = rng.choice([-1, 1], size=len(df))

    # 4. Aplicar perturbaciones
    for component in ["days", "months", "years"]:
        for sign in [-1, 1]:

            positions = np.flatnonzero(
                mask &
                (components == component) &
                (signs == sign)
            )

            if len(positions) == 0:
                continue

            out.iloc[positions] = (
                dates.iloc[positions]
                + pd.DateOffset(**{component: sign})
            )

    df[column] = out

    return df
````
